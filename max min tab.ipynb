```python
import pandas as pd

# Deskargatu fitxategia
url = "https://commons.wikimedia.org/wiki/Data:WB_WDI_SG.GEN.PARL.ZS_data.tab"
fcsv = "GEN.PARL/WB_WDI_SG.GEN.PARL.ZS_data.csv"

# Irakurri fitxategia pandas erabiliz
# df = pd.read_csv(url, sep='\t')

df = pd.read_csv(fcsv)
# hautagaiak = list(df.keys())[2:]

# Berrizendatu zutabeak ulergarriagoak izateko
df = df.rename(columns={'entity': 'herrialdea', 'year': 'urtea', 'value': 'balioa'})

# Aurkitu azken urtea
azken_urtea = df['urtea'].max()

# Iragazi azken urteko datuak
azken_urteko_datuak = df[df['urtea'] == azken_urtea]

# Iragazi herrialde jakin batzuen datuak
herrialdeak = ['USA', 'FRA', 'ESP', 'RUS', 'TUR', 'MEX', 'CHN', 'BRA', 'EGY', 'IND', 'NGA', 'ETH']
herrialde_datuak = azken_urteko_datuak[azken_urteko_datuak['herrialdea'].isin(herrialdeak)]

# Ordenatu datuak balioaren arabera
herrialde_datuak = herrialde_datuak.sort_values(by='balioa')

# Inprimatu emaitzak
print("Herrialdeen datuak, balioaren arabera ordenatuta:")
print(herrialde_datuak[['herrialdea', 'balioa']])

h10_ord =""
# for i in range(5) :
for i in herrialde_datuak['herrialdea'] :
    h10_ord = i + "," + h10_ord
print(h10_ord[:-1])
```

```python
import pandas as pd

# Deskargatu fitxategia
url = "https://commons.wikimedia.org/wiki/Data:WB_WDI_SG.GEN.PARL.ZS_data.tab"

# Irakurri fitxategia pandas erabiliz
fcsv = "GEN.PARL/WB_WDI_SG.GEN.PARL.ZS_data.csv"

df = pd.read_csv(fcsv)
# hautagaiak = list(df.keys())

# Berrizendatu zutabeak ulergarriagoak izateko
df = df.rename(columns={'entity': 'herrialdea', 'year': 'urtea', 'value': 'balioa'})

# Aurkitu azken urtea
azken_urtea = df['urtea'].max()

# Iragazi azken urteko datuak
azken_urteko_datuak = df[df['urtea'] == azken_urtea]

# Ordenatu datuak balioaren arabera
azken_urteko_datuak = azken_urteko_datuak.sort_values(by='balioa')

# Aurkitu 5 balio minimoak eta 5 balio maximoak
balio_minimoak = azken_urteko_datuak.head(5)
balio_maximoak = azken_urteko_datuak.tail(5)

# Inprimatu emaitzak
print("5 balio maximoak:")
print(balio_maximoak[['herrialdea', 'balioa']])
print()
print("5 balio minimoak:")
print(balio_minimoak[['herrialdea', 'balioa']])

h10_ord =""
# for i in range(5) :
for i in balio_minimoak['herrialdea'] :
    h10_ord = i + "," +  h10_ord 
for i in balio_maximoak['herrialdea'] :
    h10_ord = i + "," +  h10_ord

print(h10_ord[:-1])
```
